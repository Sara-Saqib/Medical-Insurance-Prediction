## Introduction

##### **Definition** A regression ML task that predicts the **individual medical insurance charges (in US dollars)** billed to a person from their personal details, such as age, body mass index (BMI), number of children, smoking status, sex and region of residence.
##### **Application** Insurance premium pricing, health-care budgeting, risk assessment by insurers, and helping individuals estimate their yearly medical costs.

### Aim of This Notebook

> **Which of 5 regression algorithms predicts medical insurance charges best, and how sure can we be about that answer?**

To answer it fairly we use **K-Fold Cross Validation** and compare the algorithms using **Root Mean Squared Error (RMSE)**.

This notebook uses the **full dataset (100 % of the records)** with **10-fold cross validation**. The five algorithms are:
**ElasticNet**, **AdaBoost**, **Histogram-Based Gradient Boosting**, **Bagging** and a **Multi-Layer Perceptron (neural network)**.

### What is K-Fold Cross Validation?

A single train/test split can be lucky or unlucky: the result depends on *which* people happened to land in the test set. K-Fold Cross Validation removes this luck:

1. Shuffle the data and split it into **K equal parts (folds)**.
2. Train the model on **K−1 folds** and test it on the **remaining fold**.
3. Repeat **K times**, so that **every fold is used as the test set exactly once**.
4. Report the **average** score over the K rounds (and the standard deviation, to show how much it varies).

With **K = 10** on the **full dataset (≈ 1,337 people)**, each test fold has about 134 people and each training set has about 1,203 people:

```
Round  1:  [TEST ] | [train] | [train] | [train] | [train] | [train] | [train] | [train] | [train] | [train]
Round  2:  [train] | [TEST ] | [train] | [train] | [train] | [train] | [train] | [train] | [train] | [train]
Round  3:  [train] | [train] | [TEST ] | [train] | [train] | [train] | [train] | [train] | [train] | [train]
Round  4:  [train] | [train] | [train] | [TEST ] | [train] | [train] | [train] | [train] | [train] | [train]
Round  5:  [train] | [train] | [train] | [train] | [TEST ] | [train] | [train] | [train] | [train] | [train]
Round  6:  [train] | [train] | [train] | [train] | [train] | [TEST ] | [train] | [train] | [train] | [train]
Round  7:  [train] | [train] | [train] | [train] | [train] | [train] | [TEST ] | [train] | [train] | [train]
Round  8:  [train] | [train] | [train] | [train] | [train] | [train] | [train] | [TEST ] | [train] | [train]
Round  9:  [train] | [train] | [train] | [train] | [train] | [train] | [train] | [train] | [TEST ] | [train]
Round 10:  [train] | [train] | [train] | [train] | [train] | [train] | [train] | [train] | [train] | [TEST ]
```

Every person is used for testing **exactly once** and for training **K−1 times**.

### Evaluation Measure: Root Mean Squared Error (RMSE)

$$RMSE = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$$

- $y_i$ = actual charges, $\hat{y}_i$ = predicted charges, $n$ = number of test people
- RMSE is in the **same unit as the output (US dollars)**, so it reads as *"on average, predictions are off by about $X"*.
- **Lower RMSE = better model.** Because errors are squared, RMSE penalises **large mistakes** heavily.

## Step 3.2: Pre-process Data

````python
"""
Pre-processing steps:
---------------------
1. Remove duplicate rows (the same person listed more than once).
   Duplicates are harmful in cross validation: a record could appear in BOTH the
   training folds and the test fold, so the model would be tested on a record it has
   already seen, and the RMSE would look better than it really is.
2. Check for missing values. This dataset has none, so nothing needs to be filled in.
   (The pipeline in Step 5 still contains imputers, so the saved model and the web
   app stay safe if a value is ever left empty.)
"""

rows_before = len(insurance)
insurance = insurance.drop_duplicates().reset_index(drop=True)
print(f"Duplicate rows removed: {rows_before - len(insurance):,}")
print(f"Instances remaining   : {len(insurance):,}")

print(f"\nMissing values after pre-processing: {insurance.isnull().sum().sum()}")
````

# Step 4: Feature Extraction

We use all **6 input features** of the dataset: 3 numerical and 3 categorical. Then we separate the **input (X)** from the **output (y)**.

````python
numeric_features = ["age", "bmi", "children"]
categorical_features = ["sex", "smoker", "region"]
target = "charges"

# Input features (X) and output (y)
X = insurance[numeric_features + categorical_features]
y = insurance[target]

print(f"Input features (X): {X.shape[1]} columns -> {list(X.columns)}")
print(f"Output (y)        : {target}")
print(f"Number of people  : {len(X):,}")

X.head()
````

# Step 5: Build the Pre-processing Pipeline

Machine learning algorithms need numbers, and some algorithms (ElasticNet, the neural network) are sensitive to the scale of each feature. The pipeline does the following:

| Feature type | Step | Why |
|---|---|---|
| Numeric (`age`, `bmi`, `children`) | **Standard scaling** (mean 0, std 1) | So that *bmi* (≈ 16 – 53) does not dominate *children* (0 – 5) |
| Categorical (`sex`, `smoker`, `region`) | **One-hot encoding** | Turns text categories into 0/1 columns. Unlike label encoding, it does not invent a fake order such as *northeast < southwest* |
| Both | **Imputation** (safety net) | Fills any empty value (median for numbers, most frequent for categories). It changes nothing here because the data has no missing values, but it protects the deployed model |

### Why put pre-processing *inside* the pipeline?

If the scaler and encoder were fitted on the **whole** dataset before splitting, the cross validation would suffer from **data leakage**: the scaler would "see" the test fold.

By placing them inside the pipeline, the cross validation loop in Step 8 **re-fits them on the training folds only** in every round, and then applies them to the test fold. This is exactly what happens in the real world, where a model never sees future data during training.

`handle_unknown="ignore"` means that if a test fold contains a category that never appeared in the training folds, the model still works and simply treats it as "no known category".

````python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features),
])

print("Pre-processing pipeline created:")
preprocessor
````

# Step 6: Define the 5 Regression Algorithms

We compare five algorithms from different families:

| # | Algorithm | How it predicts the charges |
|---|---|---|
| 1 | **ElasticNet** | A linear (weighted-sum) model that combines **L1 (Lasso)** and **L2 (Ridge)** penalties, which shrinks unimportant weights towards zero |
| 2 | **AdaBoost Regressor** | Builds many *small* decision trees one after another; each new tree pays more attention to the people the previous trees predicted badly |
| 3 | **Histogram-Based Gradient Boosting** | A fast gradient-boosting method that first groups feature values into bins (histograms) and then builds trees that correct the previous trees' errors |
| 4 | **Bagging Regressor** | Trains many decision trees on random re-samples (bootstrap samples) of the data and **averages** their predictions to reduce variance |
| 5 | **Multi-Layer Perceptron (MLP)** | A small neural network (two hidden layers of 64 and 32 neurons) that learns a non-linear function of the inputs |

All algorithms use (almost) **default settings** (no hyper-parameter tuning), so the comparison stays simple and fair. `random_state=42` makes the results reproducible.

### Why is the neural network wrapped in `TransformedTargetRegressor`?

Neural networks train badly when the **output** is a large number such as 40,000. We therefore wrap the MLP in `TransformedTargetRegressor`, which **standardises `charges` before training and converts the predictions back to dollars afterwards**. The RMSE is still measured in dollars, so it stays comparable with the other four algorithms. In our test, this one change lowered the MLP's RMSE by more than half.

The MLP is also given `max_iter=1000` (instead of the default 200) so that it has enough training rounds to converge.

````python
algorithms = {
    "ElasticNet": ElasticNet(random_state=42),

    "AdaBoost": AdaBoostRegressor(random_state=42),

    "Hist Gradient Boosting": HistGradientBoostingRegressor(random_state=42),

    "Bagging": BaggingRegressor(random_state=42, n_jobs=-1),

    "MLP Neural Network": TransformedTargetRegressor(
        regressor=MLPRegressor(hidden_layer_sizes=(64, 32), max_iter=1000, random_state=42),
        transformer=StandardScaler(),
    ),
}

print("Algorithms to compare:")
for number, name in enumerate(algorithms, start=1):
    print(f"  {number}. {name}")
````

# Step 7: Set Up 10-Fold Cross Validation
## Step 7.1: Create the K-Fold Splitter

- `n_splits=10`: split the data into 10 folds
- `shuffle=True`: shuffle the records first, so that no fold is accidentally made of one kind of person
- `random_state=42`: the **same folds are used for all 5 algorithms**. Every algorithm is tested on exactly the same people, which makes the comparison fair.

````python
K = 10
kfold = KFold(n_splits=K, shuffle=True, random_state=42)

print(f"{K}-Fold Cross Validation created (shuffle=True, random_state=42)")
````

## Step 7.2: Inspect the Folds

Let's check how many people are used for training and testing in each round.

````python
fold_sizes = []
for fold_number, (train_index, test_index) in enumerate(kfold.split(X), start=1):
    fold_sizes.append({
        "Fold (Round)": fold_number,
        "Training People": len(train_index),
        "Testing People": len(test_index),
        "Avg Charges of Test People ($)": y.iloc[test_index].mean(),
        "Highest Test Charge ($)": y.iloc[test_index].max(),
    })

fold_sizes_table = pd.DataFrame(fold_sizes).set_index("Fold (Round)")
fold_sizes_table
````

Notice that the **highest test charge** is different in each fold. A fold that happens to contain several very expensive patients will usually have a **higher RMSE**, because of the squaring explained in Step 3.1.

# Step 8: Run 10-Fold Cross Validation for All 5 Algorithms

For each algorithm we build a pipeline (**pre-processing → algorithm**). Then, for each of the 10 folds, we:

1. make a **fresh copy** of the pipeline (`clone`), so nothing learned in one round is carried into the next,
2. **train** it on the training folds,
3. **predict** the charges of the people in the test fold,
4. calculate the **RMSE** for that fold.

### Progress Bar (tqdm)

`tqdm` is a library that shows a **progress bar** while a loop runs. It is already installed in Google Colab. There is one step for each algorithm on each fold, so the bar has **5 × 10 = 50 steps**. While the cell runs, the bar looks like this:

```
Bagging, fold 2/10:  64%|██████    | 32/50 [00:35<00:20,  1.10fold/s]
```

| Part of the bar | Meaning |
|---|---|
| `Bagging, fold 2/10` | The algorithm and fold being trained right now |
| `64%` and `32/50` | How much of the work is done (steps done / total steps) |
| `00:35` (before `<`) | **Elapsed time**: time spent so far (minutes:seconds) |
| `00:20` (after `<`) | **Remaining time**: estimated time left (minutes:seconds) |
| `1.10fold/s` | Speed: folds per second (or `s/fold`, seconds per fold, when it is slow) |

> The remaining time is an **estimate** based on the average speed so far. Some algorithms are much slower than others (the neural network is the slowest), so the estimate goes up and down as each algorithm runs.
>
> ⏱ **This step takes about a minute in Colab.**

````python
fold_rmse = {}        # RMSE on each fold, for each algorithm
training_time = {}    # total time taken for all folds

folds = list(kfold.split(X))     # the same folds for every algorithm
total_steps = len(algorithms) * K

with tqdm(total=total_steps, unit="fold") as progress_bar:
    for name, algorithm in algorithms.items():
        model = Pipeline([
            ("preprocessor", preprocessor),
            ("regressor", algorithm),
        ])

        scores = []
        start = time.time()
        for fold_number, (train_index, test_index) in enumerate(folds, start=1):
            progress_bar.set_description(f"{name}, fold {fold_number}/{K}")

            fold_model = clone(model)                                   # fresh, untrained copy
            fold_model.fit(X.iloc[train_index], y.iloc[train_index])    # train on K-1 folds
            predictions = fold_model.predict(X.iloc[test_index])        # test on 1 fold
            rmse = np.sqrt(mean_squared_error(y.iloc[test_index], predictions))
            scores.append(rmse)

            progress_bar.update(1)

        training_time[name] = time.time() - start
        fold_rmse[name] = np.array(scores)
        tqdm.write(f"{name:<24} -> mean RMSE = ${fold_rmse[name].mean():>10,.2f}   ({training_time[name]:.1f} s)")

    progress_bar.set_description("Done")

print(f"\n{K}-fold cross validation completed for all {len(algorithms)} algorithms.")
````

# Step 9: Compare the Algorithms
## Step 9.1: RMSE of Each Algorithm on Each Fold

Each row is one round of cross validation, and each column is one algorithm. The **lowest RMSE in each row** (the best algorithm on that fold) is highlighted in green.

**How to read it:** count the green cells in each column to see how many folds each algorithm **won**. The algorithm with the best *average* does not always win every fold, which is why we compare the mean over all folds (Step 9.2) rather than trusting any single fold.

````python
per_fold_table = pd.DataFrame(fold_rmse)
per_fold_table.index = [f"Fold {i}" for i in range(1, K + 1)]
per_fold_table.index.name = "Fold"

(per_fold_table.style
    .format("${:,.0f}")
    .highlight_min(axis=1, props="background-color: #c8ebd2; color: #0b0b0b; font-weight: bold;"))
````

## Step 9.2: Summary Comparison Table

This is the main result of the notebook. For each algorithm:

| Column | Meaning |
|---|---|
| **Mean RMSE** | Average RMSE over the 10 folds. This is the **main measure for ranking** the algorithms (lower is better) |
| **Std RMSE** | Standard deviation over the folds. A **small value means the algorithm is consistent**; a large value means its performance depends on which people it gets |
| **Best / Worst Fold** | Lowest and highest RMSE across the folds |
| **RMSE as % of Avg Charges** | Mean RMSE divided by the average charges. This puts the error in context |
| **Folds Won** | In how many of the 10 folds this algorithm had the lowest RMSE |
| **Time** | Total time to train and test across all 10 folds |
| **Rank** | 1 = lowest mean RMSE = best algorithm |

````python
average_charges = y.mean()

summary = pd.DataFrame({
    "Mean RMSE ($)": per_fold_table.mean(),
    "Std RMSE ($)": per_fold_table.std(),
    "Best Fold RMSE ($)": per_fold_table.min(),
    "Worst Fold RMSE ($)": per_fold_table.max(),
    "RMSE as % of Avg Charges": per_fold_table.mean() / average_charges * 100,
    "Folds Won": per_fold_table.idxmin(axis=1).value_counts().reindex(per_fold_table.columns).fillna(0).astype(int),
    "Time (s)": pd.Series(training_time),
})
summary = summary.sort_values("Mean RMSE ($)")
summary.insert(0, "Rank", range(1, len(summary) + 1))
summary.index.name = "Algorithm"

print(f"Average medical charges in this data: ${average_charges:,.2f}\n")

money_columns = ["Mean RMSE ($)", "Std RMSE ($)", "Best Fold RMSE ($)", "Worst Fold RMSE ($)"]
(summary.style
    .format({**{c: "${:,.0f}" for c in money_columns},
             "RMSE as % of Avg Charges": "{:.1f}%",
             "Time (s)": "{:.2f}"})
    .highlight_min(subset=["Mean RMSE ($)"], props="background-color: #c8ebd2; color: #0b0b0b; font-weight: bold;"))
````

## Step 9.3: Visual Comparison

**Chart 1** shows the mean RMSE of each algorithm (shorter bar = better). The black whisker shows ± 1 standard deviation across the 10 folds.

**Chart 2** shows the RMSE of **every individual fold** as a dot, so you can see how spread out each algorithm's results are.

````python
# Chart colours: one blue for the data, dark/grey text, light grid
BAR_COLOR = "#2a78d6"
TEXT_COLOR = "#0b0b0b"
MUTED_TEXT = "#52514e"
GRID_COLOR = "#e4e3df"
dollars = FuncFormatter(lambda value, _: f"${value:,.0f}")

def style_axis(ax):
    for side in ["top", "right", "left"]:
        ax.spines[side].set_visible(False)
    ax.spines["bottom"].set_color(GRID_COLOR)
    ax.tick_params(colors=MUTED_TEXT, length=0)
    ax.xaxis.set_major_formatter(dollars)
    ax.xaxis.set_major_locator(MaxNLocator(nbins=6))
    ax.grid(axis="x", color=GRID_COLOR, linewidth=0.8)
    ax.set_axisbelow(True)

order = summary.index.tolist()            # best algorithm first
positions = np.arange(len(order))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.2))

# ---- Chart 1: mean RMSE with +/- 1 std ----
means = summary["Mean RMSE ($)"]
stds = summary["Std RMSE ($)"]
ax1.barh(positions, means, height=0.5, color=BAR_COLOR,
         xerr=stds, error_kw=dict(ecolor=TEXT_COLOR, elinewidth=1.2, capsize=4))
for i, name in enumerate(order):
    label = f"${means[name]:,.0f}" + ("   (best)" if i == 0 else "")
    ax1.text(means[name] + stds[name] + means.max() * 0.02, i, label,
             va="center", color=TEXT_COLOR, fontsize=10)
ax1.set_yticks(positions, order, color=TEXT_COLOR)
ax1.invert_yaxis()
ax1.set_xlim(0, (means + stds).max() * 1.3)
style_axis(ax1)
ax1.set_title("Mean RMSE across 10 folds (lower is better)", loc="left", color=TEXT_COLOR, fontsize=12)
ax1.set_xlabel("RMSE (US dollars)", color=MUTED_TEXT)

# ---- Chart 2: RMSE of every fold ----
for i, name in enumerate(order):
    ax2.scatter(per_fold_table[name], np.full(K, i), s=60, color=BAR_COLOR,
                edgecolor="white", linewidth=1.5, zorder=3,
                label="RMSE of one fold" if i == 0 else None)
    ax2.scatter(means[name], i, marker="|", s=500, color=TEXT_COLOR, linewidth=2.5, zorder=4,
                label="Mean RMSE" if i == 0 else None)
ax2.set_yticks(positions, order, color=TEXT_COLOR)
ax2.invert_yaxis()
ax2.set_xlim(0, per_fold_table.values.max() * 1.1)
style_axis(ax2)
ax2.set_title("RMSE of each fold (spread = consistency)", loc="left", color=TEXT_COLOR, fontsize=12)
ax2.set_xlabel("RMSE (US dollars)", color=MUTED_TEXT)
ax2.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.16), ncol=2, labelcolor=MUTED_TEXT)

plt.tight_layout()
plt.show()
````

## Step 9.4: Observations

````python
# These observations are generated from YOUR run, so the numbers always match the tables above.
best, second, worst = summary.index[0], summary.index[1], summary.index[-1]
gap = summary.loc[second, "Mean RMSE ($)"] - summary.loc[best, "Mean RMSE ($)"]
best_std = summary.loc[best, "Std RMSE ($)"]
linear_rmse = summary.loc["ElasticNet", "Mean RMSE ($)"]
fastest = summary["Time (s)"].idxmin()
slowest = summary["Time (s)"].idxmax()

print("Observations from this run")
print("==========================\n")
print(f"1. Best algorithm : {best} (mean RMSE ${summary.loc[best, 'Mean RMSE ($)']:,.0f}, "
      f"{summary.loc[best, 'RMSE as % of Avg Charges']:.1f}% of the average charge).")
print(f"2. Runner-up      : {second} is only ${gap:,.0f} behind. The best algorithm's fold-to-fold "
      f"standard deviation is ${best_std:,.0f},")
print(f"                    so " + ("the two are practically TIED." if gap < best_std else
                                    "the gap is larger than the fold-to-fold variation."))
print(f"3. Worst algorithm: {worst} (mean RMSE ${summary.loc[worst, 'Mean RMSE ($)']:,.0f}).")
print(f"4. ElasticNet is a linear model (RMSE ${linear_rmse:,.0f}); the best model reduces its error by "
      f"{(1 - summary.loc[best, 'Mean RMSE ($)'] / linear_rmse) * 100:.0f}%.")
print(f"5. Speed: {fastest} is the fastest ({summary.loc[fastest, 'Time (s)']:.1f} s) and "
      f"{slowest} is the slowest ({summary.loc[slowest, 'Time (s)']:.1f} s) for all 10 folds.")
````

**How to interpret these results (general points):**

1. **Non-linear models usually beat a linear model on this data.** Medical charges do not grow in a straight line with the features: being a **smoker** *and* having a **high BMI** raises charges far more than the two effects added together (an *interaction*). Tree-based and neural-network models can learn such interactions on their own, whereas ElasticNet (a weighted sum) cannot.
2. **Small RMSE differences between the top algorithms may be noise.** Always compare the gap with the **standard deviation across folds** before declaring a clear winner.
3. **RMSE is in dollars**, so it can be explained to a non-technical user: *"a typical prediction is off by about this many dollars."*
4. **Error depends on the fold.** A fold with several very expensive patients gives a higher RMSE for **every** algorithm (see the per-fold table in Step 9.1).
5. **Running 10 folds gives a more reliable comparison than a single train/test split**, because each algorithm is tested 10 times and every person is used for testing exactly once.

# Step 12: Execute the Feedback Phase

After deployment, feedback from users and domain experts (e.g. insurance analysts) is used to improve the model. Potential improvements for this system:

1. **Add more features**: pre-existing conditions, family medical history, occupation, income and exercise habits strongly affect real insurance costs but are not in this dataset.
2. **Tune hyper-parameters**: e.g. the learning rate and number of iterations of Histogram Gradient Boosting, the layer sizes of the MLP, or the `l1_ratio` of ElasticNet. Use cross validation (`GridSearchCV`) to choose them.
3. **Handle the skewed charges**: report an error measure that is less sensitive to outliers (e.g. **MAE**) next to RMSE, or train separate models for smokers and non-smokers.
4. **Add feature interactions**: explicitly create features such as `smoker × bmi` to help linear models such as ElasticNet.
5. **Collect more and more recent data**: the dataset has only about 1,300 people, and costs change from year to year.
6. **Report uncertainty**: show a prediction *range* (e.g. ± RMSE) in the web app instead of a single number, and tell users the estimate is **not** an official insurance quotation.
